```python
#1부터 100까지의 합을 출력하시오.
sum=0
for i in range(1,101):
  sum=sum+i
print(sum)
```

```python
#1부터 100까지의 짝수의 합을 range()의 증가값을 이용하여 구하세요. (if문 미사용)
sum = 0
for a in range(2, 101, 2):
  sum=sum+a
print(sum)
```

```python
#1부터 100까지의 짝수의 합을 range()의 증가값을 이용하여 구하세요. (if문 사용)
sum= 0
for i in range(1, 101):
 if i % 2 == 0:
  sum=sum+i
print(sum)
```

```python
#정수 하나를 입력받아 그 수의 구구단을 7 x 1 = 7 형식으로 출력하세요.
n=int(input("단의 정수를 입력하세요."))
for i in range(1, 10):
   print(n, 'x', '=', n*i )
```

```python
#1에서 100까지 더하시오. (while문으로)
sum=0
i=1
while i<=100:
  sum+=i
  i+=1
print(sum)
```

```python
sum=0
i=1
while True:
  sum+=i
  i+=1
  if i>100:
    break
print(sum)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요.
sum=0
while True:
  n=int(input("양의정수를 입력하세요."))
  if n==0:
    break
  sum+=n
print(sum)
```

```python
#1부터 20까지의 수 중 짝수만 담은 리스트를 만들어 출력하세요.
evens=[]
for i in range(1,21):
  if i%2==0:
    evens.append(i)
print(evens)
evens.pop(7)
print(evens)
del evens[5]
print(evens)
```
