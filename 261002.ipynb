```python
def add(n1, n2):
    a = random.randint(1, n1)
    b = random.randint(1, n2)
    result = a + b
    print(f'{a} + {b} = {result}')
```

```python
def asmd(x,y,op):
    if op=='+':
      print(f'{x} + {y} = {add(x,y)})')
    elif op=='-':
      print(f'{x} - {y} = {add(x,y)})')
    elif op=='*':
      print(f'{x} * {y} = {add(x,y)})')
    elif op=='/':
      print(f'{x} / {y} = {add(x,y)}:.2f)')
```

```python
x = int(input('첫 번째 수: '))
y = int(input('두 번째 수: '))
op=input('연산자를 입력하시오')
asmd(x,y,op)
```

```python
import random
com = random.randint(1, 100)
count=0
while True:
   user = int(input('1에서 100까지의 숫자 중 맞혀보세요.: '))
   count+=1
   if com == user:
     print('정답입니다.')
     print(count, '번 만에 맞혔습니다.')
     break
   elif com > user:
     print('컴퓨터의 숫자가 더 큽니다.')
   else:
     print('컴퓨터의 숫자가 더 작습니다.')
```
